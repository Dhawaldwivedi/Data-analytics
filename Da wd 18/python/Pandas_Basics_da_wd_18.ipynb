# Pandas

```python
# data manipulation
import numpy as np
import pandas as pd

```

```python
lst = [1,2,3,4,5]
arr = np.array(lst)
print(arr)
print(type(arr))
```

```python
s = pd.Series(arr)
s
```

```python
# creating a series from a dictionary
dct = {'a':'hi','b':'hey','c':'hello'}
s = pd.Series(dct)
s
```

```python
# creating a series from lists
lst1 = [1,4,6,8]
lst2 = ['hey','hi','gfg','hello']
s = pd.Series(lst2 , index = lst1)
s
```

```python
# dataframe
dct = {'student1':{'name':'Dhawal','age':20,'sem':3},
       'student2':{'name':'Aman','age':30,'sem':6},
       'student3':{'name':'Ansh','age':24,'sem':4}}
df = pd.DataFrame(dct)
df
```

```python
df['student3']
```

```python
# loc - label based indexing
df.loc['age']['student2']
```

```python
# iloc - integer based indexing
print(df.iloc[2][1])
```

```python
# adding a column
df['student 4'] = ['Akash',23,1]
df
```

```python
# drop a column
df.drop('student 4',axis = 1 , inplace = True) # removing column row wise
df
```

```python
df.drop('age',axis=0,inplace = True) # removing row column wise
df
```

```python
# revenue per region
revenue_by_region = df.groupby('Region')['Total Revenue'].sum().sort_values(ascending = False)
revenue_by_region
```

```python
# revenue per payment method
revenue_by_paymentmethod = df.groupby('Payment Method')['Total Revenue'].sum().sort_values(ascending = False)
revenue_by_paymentmethod
```
