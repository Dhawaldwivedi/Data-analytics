```python
# wap to check if a number is prime or not
n = 5
count = 0
for i in range(2,n):
  if n%i ==0:
    print('not prime')
    break
  else :
    count++
    print('prime')

```

```python
def pattern(n):
  for i in range(n):
    for j in range(n):
      print('*', end = "")
    print()
```
