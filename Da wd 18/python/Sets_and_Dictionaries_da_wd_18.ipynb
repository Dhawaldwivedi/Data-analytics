**count each number occurrence in a list**  

```python
lst = [1,2,3,4,2,4,5,6,5,3,7,89]
freq = {}

for i in lst:               # {1:1,2:2,3:1,4:2,  }
  if i in freq:
    freq[i] = freq[i]+1
  else:
    freq[i] = 1

print(freq)
```

**Unique Common Elements Problem: You are given two lists of integers. Write a program to:
