```python
import numpy as np
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
# line plot
x = [2,3,4,5,6,7]
y = [10,20,30,40,50,60]
plt.figure(figsize=(4,4))
plt.plot(x,y,linewidth = 2, marker = 'o',color = 'c')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Plot of X and Y')
plt.show()
```

```python
print(plt.style.available)
```

```python
plt.style.use("dark_background")
x = [2,3,4,5,6,7]
y = [10,20,30,40,50,60]
plt.figure(figsize=(4,4))
plt.plot(x,y,linewidth = 2, marker = 'o',color = 'c')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Plot of X and Y')
plt.show()
```

```python
# bar plot - plt.bar()
x= ["A","B","C","D"]
y = [10,65,35,20]
plt.figure(figsize=(4,4))
plt.bar(x,y)
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

```python
# scatter plot
import numpy as np
x= np.random.randint(1,50,20)
y = np.random.randint(1,50,20)
plt.figure(figsize=(4,4))
plt.scatter(x,y)
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

```python
# plt.hist (histogram)
x= np.random.rand(50)
plt.hist(x,bins=10)
plt.show()
```
